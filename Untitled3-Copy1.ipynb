```python
import torch
import numpy as np
from torch.nn import *
import matplotlib.pyplot as plt
from PIL.Image import fromarray
from tqdm.notebook import tqdm


class F(Module):
    def __init__(s, f):
        super().__init__()
        s.f = f

    def forward(s, x):
        return s.f(x)


a = torch.stack(
    torch.broadcast_tensors(
        torch.linspace(-1, 1, 1080)[:, None], torch.linspace(-1, 1, 1920)
    ),
    2,
)
cm = plt.cm.rainbow(np.linspace(0, 1, 16))[:, :3] * 255.999
m = Sequential(Linear(2, 4), Hardswish(1), Linear(4, len(cm)))
o = torch.optim.Adam(m.parameters(), 0.2)
for _ in range(20):
    for _ in tqdm(range(20), leave=0):
        _ = m(a).softmax(2)
        (_.mean((0, 1)).std() - _.std(2).mean()).backward()
        o.step()
        o.zero_grad()
    display(fromarray((_.detach().numpy() @ cm).astype("B")))
```
