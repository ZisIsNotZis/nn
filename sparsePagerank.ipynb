```python
from scipy.sparse import coo_matrix
import numpy as np

a = coo_matrix(
    (
        np.random.rand(1048576 * 3).astype("f"),
        (
            np.arange(1048576 * 3, dtype="I") % 1048576,
            np.random.randint(1048576, size=1048576 * 3).astype("I"),
        ),
    ),
    (1048576, 1048576),
).tocsr()
```

```python
from multiprocessing.pool import ThreadPool


def f(i, v):
    a[i] /= v.sum()


with ThreadPool(8) as p:
    p.starmap(f, enumerate(a))
```

```python
k = np.ones(1048576, 'f')
```

```python
for _ in range(20):
    k2 = k @ a
    print(np.linalg.norm(k2-k))
    k = k2
```

```python
import matplotlib.pyplot as plt
plt.hist(k)
plt.yscale('log');
```
