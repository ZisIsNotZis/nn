```python
from PIL.Image import open, fromarray
import numpy as np
import torch
from torch.nn import *

torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = torch.tensor(np.array(open("a.png"), "f").mean(2) / 255)
f = (
    torch.mean,
    lambda x, i: x.square().mean(i),
    lambda x, i: x.sqrt().mean(i),
    torch.std,
    lambda x, i: x.square().std(i),
    lambda x, i: x.sqrt().std(i),
    lambda x, i: x.diff(dim=i).square_().mean(i),
    lambda x, i: x.diff(dim=i).diff(dim=i).square_().mean(i),
)
k = [[j(a, i) for j in f] for i in range(2)]
```

```python
b = torch.rand((1080, 1920), requires_grad=True)
c = MSELoss()
o = torch.optim.Adam((b,), 0.01)
for _ in range(10):
    for _ in range(500):
        _ = b.sigmoid()
        _ = sum(c(j(_, i), k) for i, k in enumerate(k) for j, k in zip(f, k))
        (_ * 65536.0).backward()
        o.step()
        o.zero_grad()
    print(_)
```
