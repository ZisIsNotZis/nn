```python
import torch
import numpy as np
from torch.nn import *

x = torch.stack(
    torch.broadcast_tensors(torch.arange(0, 1, 0.01)[:, None], torch.arange(0, 1, 0.01))
).view(-1, 2)
y = torch.ones(10)
y /= y.sum()
m = Sequential(Linear(2, 4), Hardswish(1), Linear(4, len(y)), Softmax(1))
c = MSELoss()
o = torch.optim.Adam(m.parameters(), 0.1)
for _ in range(10):
    for _ in range(100):
        _ = m(x)
        _ = c(_, y) - _.std(0).mean()
        _.backward()
        o.step()
        o.zero_grad()
    print(_)
```

```python
import matplotlib.pyplot as plt

with torch.no_grad():
    plt.imshow(m(x).argmax(1).view(int(len(x) ** 0.5), -1))
```
