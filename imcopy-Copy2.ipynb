```python
import torch
import numpy as np
from torch.nn import *
from PIL.Image import open, fromarray

torch.set_default_tensor_type(torch.cuda.FloatTensor)


class F(Module):
    def __init__(s, f):
        super().__init__()
        s.f = f

    def forward(s, x):
        return s.f(x)


x = torch.stack(torch.meshgrid(torch.linspace(-1, 1, 54), torch.linspace(-1, 1, 91)), 2)
y = open("assets/a.png").resize((91, 54), 2)
display(y)
y = torch.tensor(np.array(y), dtype=torch.float).div_(255.)
m = Sequential(Linear(2, 256), F(torch.sinh_), Linear(256, 3))
c = MSELoss()
o = torch.optim.Adam(m.parameters(), 1e-5)
while 1:
    for _ in range(10):
        for _ in range(100000):
            _ = m(x)
            l = c(_, y).mul_(65536.0)
            l.backward()
            o.step()
            o.zero_grad()
        print(l)
    display(fromarray(_.detach().mul_(255.).to("cpu", torch.uint8).numpy()))
```

```python
torch.save((m,o), 'imcopy2.mo')
```
