```python
import torch
import numpy as np
from torch.nn import *
from PIL.Image import open, fromarray, LANCZOS


class F(Module):
    def __init__(s, f):
        super().__init__()
        s.f = f

    def forward(s, x):
        return s.f(x)


torch.set_default_tensor_type(torch.cuda.FloatTensor)
x = torch.stack(
    torch.broadcast_tensors(
        torch.linspace(-1, 1, 54)[:, None], torch.linspace(-1, 1, 91)
    ),
    2,
)
y = torch.tensor(np.array(open("assets/a.png").resize((91, 54), LANCZOS))).to(
    "cuda", torch.float
)
m = Sequential(Linear(2, 1024), F(torch.sinh_), Linear(1024, 3))
c = MSELoss()
o = torch.optim.Adam(m.parameters(), 1e-4)
for _ in range(100):
    for _ in range(20):
        for _ in range(1000):
            _ = m(x).sigmoid_() * 255
            __ = c(_, y)
            __.backward()
            o.step()
            o.zero_grad()
        print(__)
    display(fromarray((_.detach().to("cpu", torch.uint8).numpy())))
```
