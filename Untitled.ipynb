```python
def f(n, b=0):
    if n:
        if b:
            for i in f(n-1, b-1):
                yield ')' + i
        if b < n:
            for i in f(n-1, b+1):
                yield '(' + i
    else:
        yield ''
def F(n):
    for i in range(0, n+1, 2):
        yield from f(i)
```

```python
from pyparsing import *

atom = (
    quotedString().setParseAction(lambda x: x[0][1:-1])
    ^ Word(nums + "-").setParseAction(lambda x: int(x[0]))
    ^ Word(nums + ".-").setParseAction(lambda x: float(x[0]))
    ^ Or("true", "false").setParseAction(lambda x: bool(x))
    ^ Literal("null").setParseAction(lambda: ...)
    ^ Word(alphanums)
)
dict = Forward()
list = Forward()
obj = atom ^ Suppress("{") + dict + Suppress("}") ^ Suppress("[") + list + Suppress("]")
dict <<= Group(delimitedList(obj + Suppress(":") + obj))
list <<= Group(delimitedList(obj ^ dict))
json = obj ^ dict ^ list
```
